# Part-2: Working with Model

# 1) Create the target data and feature data where target data is price

```python
#1) Create the target data and feature data where target data is price
```

```python
x = df.drop('Price' , axis = 1)
y = df['Price']
```

```python
import matplotlib.pyplot as plt
```

```python
print(len(x))
print(len(y))
```

```python
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.8,random_state=)
```

# 2) Create a linear regression model for Target and feature data

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

# 3) Check if the model is overfitting or underfitting or it is accurate

```python
y_pred = model.predict(x_test)
```

```python
model.score(x_train, y_train)*100
```

```python
model.score(x_test, y_test)*100
```

# 5) Extract slope and intercept value from the model

```python
print(model.intercept_)
```

```python
print(model.coef_)
```

```python
coef = pd.DataFrame({
    'Feature': x.columns,
    'Coef': model.coef_
})

coef.sort_values(
    by='Coef',
    ascending=False
)
```
